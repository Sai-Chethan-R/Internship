```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Environment Setup & Data Generation
# Creating a synthetic dataset for the lab exercise
np.random.seed(42)
dates = pd.date_range(start='2026-01-01', periods=100, freq='D')
data = {
    'Date': dates,
    'Category': np.random.choice(['Electronics', 'Clothing', 'Home', 'Toys'], 100),
    'Sales_Price': np.random.uniform(50, 500, 100),
    'Discount': np.random.uniform(0, 0.3, 100),
    'Units_Sold': np.random.randint(1, 20, 100)
}
df = pd.DataFrame(data)

# Introduce missing values to demonstrate data cleaning
df.loc[5:10, 'Sales_Price'] = np.nan

# 2. Data Manipulation & Cleaning
print("Missing values before cleaning:\n", df.isnull().sum())

# Handle missing values by imputing the mean
df['Sales_Price'] = df['Sales_Price'].fillna(df['Sales_Price'].mean())

# Feature engineering: Calculate total revenue per transaction
df['Total_Revenue'] = df['Sales_Price'] * df['Units_Sold'] * (1 - df['Discount'])

# Grouping and aggregation
category_summary = df.groupby('Category').agg({
    'Total_Revenue': 'sum',
    'Units_Sold': 'sum'
}).reset_index()

print("\n--- Category Summary ---")
print(category_summary)

# 3. Basic Analysis & Visualization
sns.set_theme(style="whitegrid")

# Visualization 1: Bar Chart of Revenue by Category
plt.figure(figsize=(8, 5))
sns.barplot(x='Category', y='Total_Revenue', data=category_summary, palette='viridis')
plt.title('Total Revenue by Product Category')
plt.ylabel('Total Revenue ($)')
plt.xlabel('Category')
plt.tight_layout()
plt.savefig('revenue_by_category.png')
plt.show()

# Visualization 2: Line Chart of Daily Revenue Trend
plt.figure(figsize=(10, 5))
sns.lineplot(x='Date', y='Total_Revenue', data=df, color='coral')
plt.title('Daily Revenue Trend over 100 Days')
plt.ylabel('Revenue ($)')
plt.xlabel('Date')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('daily_revenue_trend.png')
plt.show()
```
